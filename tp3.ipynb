#### Configuration et Imports

```python
import pandas as pd
import numpy as np

DATA_YELP = 'data/use_case'
DATASET='data/custom_dataset'

VILLES = ['Phoenix', 'Toronto', 'Las Vegas']

```

#### Chargement des données

```python
business_df = pd.read_csv(DATA_YELP + '/restaurants.csv')
avis_df = pd.read_csv(DATA_YELP+'/avis.csv')
user_df = pd.read_csv(DATA_YELP+'/utilisateurs.csv')
checkin_df = pd.read_csv(DATA_YELP+'/checkin.csv')
conseils_df = pd.read_csv(DATA_YELP+'/conseils.csv')
horaires_df = pd.read_csv(DATA_YELP+'/horaires.csv')

print("Restaurants : " + str(business_df.shape))
print('Avis : ' + str(avis_df.shape))
print('Utilisateurs : ' +str(user_df.shape))
print('Checking : ' +str(checkin_df.shape))
print('Conseils : ' +str(conseils_df.shape))
print('Horaires : ' +str(horaires_df.shape))
```

#### Colonnes

```python
cl_business = business_df[['restaurant_id','nom','ville','moyenne_etoiles']]
cl_avis = avis_df[['restaurant_id','utilisateur_id','etoiles','date']]
cl_users= user_df[['utilisateur_id','nom','elite']]
cl_checkin= checkin_df[['restaurant_id','date']]
cl_conseils= conseils_df[['restaurant_id','utilisateur_id']]
cl_horaire= horaires_df
```

#### Filtrage

```python
cl_business = cl_business[cl_business['ville'].isin(VILLES)]
restaurant_ids = cl_business['restaurant_id']

cl_avis = cl_avis[cl_avis['restaurant_id'].isin(restaurant_ids)]
cl_checkin = cl_checkin[cl_checkin['restaurant_id'].isin(restaurant_ids)]
cl_conseils =cl_conseils[cl_conseils['restaurant_id'].isin(restaurant_ids)]
cl_horaire=cl_horaire[cl_horaire['restaurant_id'].isin(restaurant_ids)]

print("Avis :"+ str(cl_avis))
```

#### Empty Values

```python
cl_business= cl_business.dropna()
cl_avis= cl_avis.dropna()
cl_users = cl_users.dropna()
cl_checkin = cl_checkin.dropna()
cl_conseils = cl_conseils.dropna()
cl_horaire = cl_horaire.dropna()

print(cl_business.isnull().sum().sum())
print(cl_avis.isnull().sum().sum())
print(cl_users.isnull().sum().sum())
print(cl_checkin.isnull().sum().sum())
print(cl_conseils.isnull().sum().sum())
print(cl_horaire.isnull().sum().sum())
```

#### Sauvegarde des données

```python
cl_business.head(5000).to_csv(DATASET + '/restaurants.csv',index=False)
cl_avis.head(5000).to_csv(DATASET + '/avis.csv',index=False)
cl_users.head(5000).to_csv(DATASET + '/users.csv',index=False)
cl_checkin.head(5000).to_csv(DATASET + '/checkin.csv',index=False)
cl_conseils.head(5000).to_csv(DATASET + '/conseils.csv',index=False)
cl_horaire.head(5000).to_csv(DATASET + '/horaires.csv',index=False)


```
